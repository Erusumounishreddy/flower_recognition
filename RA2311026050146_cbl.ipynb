```python
!pip -q install scikit-learn seaborn pandas matplotlib

import os, time, copy, json, random, shutil
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from PIL import Image
import torch, torch.nn as nn, torchvision
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms
from torchvision.datasets import Flowers102
from sklearn.metrics import classification_report, confusion_matrix, f1_score

SEED, IMG, SRC, BS, NC = 42, 224, 256, 32, 102
OUT = "/content/PS38_outputs"; os.makedirs(OUT, exist_ok=True)
dev = torch.device("cuda" if torch.cuda.is_available() else "cpu")
def seed_all(s=SEED):
    random.seed(s); np.random.seed(s); torch.manual_seed(s); torch.cuda.manual_seed_all(s)
seed_all()
print("Device:", dev, "| torch", torch.__version__, "| torchvision", torchvision.__version__)

# ---------- data (official split: 10 train / 10 val per class) ----------
raw = {s: Flowers102("/content/data", split=s, download=True) for s in ["train", "val", "test"]}
def load(s, size):
    X, y = [], []
    for im, l in raw[s]: X.append(im.convert("RGB").resize((size, size), Image.BILINEAR)); y.append(int(l))
    return X, np.array(y)
Xtr, ytr = load("train", SRC); Xva, yva = load("val", IMG); Xte, yte = load("test", IMG)
print("Split ->", len(Xtr), len(Xva), len(Xte))
ctr, cva, cte = [np.bincount(y, minlength=NC) for y in (ytr, yva, yte)]
print("Per class: train", ctr.min(), "-", ctr.max(), "| val", cva.min(), "-", cva.max(), "| test", cte.min(), "-", cte.max())

plt.figure(figsize=(12, 3)); plt.bar(range(NC), cte, color="tab:green", label="test")
plt.axhline(10, c="tab:blue", ls="--", label="train/val = 10 per class"); plt.legend()
plt.xlabel("class index"); plt.ylabel("images"); plt.title("Images per class")
plt.tight_layout(); plt.savefig(f"{OUT}/fig1_class_distribution.png", dpi=200); plt.show()
fig, axs = plt.subplots(2, 8, figsize=(16, 4.5))
for a, c in zip(axs.ravel(), np.random.RandomState(SEED).choice(NC, 16, replace=False)):
    a.imshow(Xtr[int(np.where(ytr == c)[0][0])]); a.set_title(f"class {c}", fontsize=8); a.axis("off")
plt.tight_layout(); plt.savefig(f"{OUT}/fig2_samples.png", dpi=200); plt.show()

M, S = [0.485, 0.456, 0.406], [0.229, 0.224, 0.225]
tf_aug = transforms.Compose([transforms.RandomResizedCrop(IMG, scale=(0.5, 1.0)), transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20), transforms.ColorJitter(0.2, 0.2, 0.2), transforms.ToTensor(), transforms.Normalize(M, S)])
tf_tr_plain = transforms.Compose([transforms.Resize((IMG, IMG)), transforms.ToTensor(), transforms.Normalize(M, S)])
tf_plain = transforms.Compose([transforms.ToTensor(), transforms.Normalize(M, S)])   # val/test: no augmentation
class DS(Dataset):
    def __init__(s, X, y, tf): s.X, s.y, s.tf = X, y, tf
    def __len__(s): return len(s.X)
    def __getitem__(s, i): return s.tf(s.X[i]), int(s.y[i])
g = torch.Generator(); g.manual_seed(SEED)
dl_aug = DataLoader(DS(Xtr, ytr, tf_aug), BS, shuffle=True, num_workers=2, generator=g)
dl_plain = DataLoader(DS(Xtr, ytr, tf_tr_plain), BS, shuffle=True, num_workers=2, generator=g)
dl_val = DataLoader(DS(Xva, yva, tf_plain), 128, num_workers=2)
dl_test = DataLoader(DS(Xte, yte, tf_plain), 128, num_workers=2)

# ---------- model + training ----------
def build():
    m = models.efficientnet_b0(weights=models.EfficientNet_B0_Weights.IMAGENET1K_V1)
    m.classifier[1] = nn.Linear(1280, NC); return m
ce, ce_s = nn.CrossEntropyLoss(), nn.CrossEntropyLoss(label_smoothing=0.1)

@torch.no_grad()
def evaluate(m, dl):
    m.eval(); tl = n = 0; P, L, Z = [], [], []
    for x, y in dl:
        x, y = x.to(dev), y.to(dev); z = m(x); tl += ce(z, y).item() * len(y); n += len(y)
        P.append(z.argmax(1).cpu()); L.append(y.cpu()); Z.append(z.cpu())
    P, L, Z = torch.cat(P).numpy(), torch.cat(L).numpy(), torch.cat(Z)
    return tl / n, (P == L).mean(), P, L, Z

def fit(m, dl, lr, ep, pat, h, crit):
    opt = torch.optim.AdamW([p for p in m.parameters() if p.requires_grad], lr=lr, weight_decay=1e-4)
    sch = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=ep)
    best, bs, bad = 1e9, copy.deepcopy(m.state_dict()), 0
    for e in range(ep):
        m.train(); tl = tc = tn = 0
        for x, y in dl:
            x, y = x.to(dev), y.to(dev); opt.zero_grad(); o = m(x); l = crit(o, y); l.backward(); opt.step()
            tl += l.item() * len(y); tc += (o.argmax(1) == y).sum().item(); tn += len(y)
        sch.step(); vl, va, *_ = evaluate(m, dl_val)
        h["train_loss"].append(tl/tn); h["train_acc"].append(tc/tn); h["val_loss"].append(vl); h["val_acc"].append(float(va))
        print(f"  ep {len(h['val_acc']):02d} | train {tc/tn:.3f}/{tl/tn:.3f} | val {va:.3f}/{vl:.3f}")
        if vl < best - 1e-4: best, bs, bad = vl, copy.deepcopy(m.state_dict()), 0
        else:
            bad += 1
            if bad >= pat: break
    m.load_state_dict(bs)

def run(name, dl, crit):
    print(f"\n=== {name} ==="); seed_all(); m = build().to(dev); h = dict(train_loss=[], train_acc=[], val_loss=[], val_acc=[]); t0 = time.time()
    for p in m.features.parameters(): p.requires_grad = False
    fit(m, dl, 1e-3, 5, 99, h, crit)                     # stage 1: head only
    for p in m.parameters(): p.requires_grad = True
    fit(m, dl, 1e-4, 15, 4, h, crit)                     # stage 2: fine-tune all layers
    return m, h, time.time() - t0

mods, hist, tm = {}, {}, {}
mods["B"], hist["B"], tm["B"] = run("B: EffNet-B0, no augmentation", dl_plain, ce)
mods["C"], hist["C"], tm["C"] = run("C: EffNet-B0 + augmentation (final)", dl_aug, ce_s)

# ---------- learning curves ----------
fig, ax = plt.subplots(1, 2, figsize=(12, 4)); col = {"B": "tab:blue", "C": "tab:red"}
for k, h in hist.items():
    e = range(1, len(h["val_acc"]) + 1)
    for a, t in zip(ax, ["acc", "loss"]):
        a.plot(e, h[f"train_{t}"], "--", c=col[k], label=f"{k} train"); a.plot(e, h[f"val_{t}"], c=col[k], label=f"{k} val")
for a, t in zip(ax, ["Accuracy", "Loss"]): a.set_title(t); a.set_xlabel("Epoch"); a.grid(alpha=.3); a.legend()
plt.tight_layout(); plt.savefig(f"{OUT}/fig3_learning_curves.png", dpi=200); plt.show()

# ---------- test evaluation (once per model) ----------
def latency(m):
    m.eval(); x = torch.randn(1, 3, IMG, IMG, device=dev)
    with torch.no_grad():
        for _ in range(10): m(x)
        if dev.type == "cuda": torch.cuda.synchronize()
        t = time.time()
        for _ in range(50): m(x)
        if dev.type == "cuda": torch.cuda.synchronize()
    return (time.time() - t) / 50 * 1000

res, out = {}, {}
for k, m in mods.items():
    tl, ta, P, L, Z = evaluate(m, dl_test); out[k] = (P, L, Z); h = hist[k]
    rep = classification_report(L, P, labels=list(range(NC)), output_dict=True, zero_division=0)
    pd.DataFrame(rep).T.round(4).to_csv(f"{OUT}/per_class_report_{k}.csv")
    rec = np.array([rep[str(c)]["recall"] for c in range(NC)])
    res[k] = dict(test_acc=float(ta), top5_acc=float((Z.topk(5, 1).indices.numpy() == L[:, None]).any(1).mean()),
        macro_f1=float(f1_score(L, P, average="macro")), macro_precision=rep["macro avg"]["precision"],
        macro_recall=rep["macro avg"]["recall"], min_class_recall=float(rec.min()), classes_recall_below_80=int((rec < .8).sum()),
        params=sum(p.numel() for p in m.parameters()), epochs=len(h["val_acc"]), train_time_s=tm[k], latency_ms=latency(m),
        last_train_acc=h["train_acc"][-1], last_val_acc=h["val_acc"][-1], last_train_loss=h["train_loss"][-1], last_val_loss=h["val_loss"][-1])
    print(k, {a: round(b, 4) for a, b in res[k].items()})
P, L, Z = out["C"]; p = res["C"]["test_acc"]; res["C"]["margin_of_error_95"] = float(1.96 * np.sqrt(p * (1 - p) / len(L)))
pd.DataFrame(res).T.round(4).to_csv(f"{OUT}/table_overall_results.csv")

# ---------- confusion matrix, confused pairs, worst classes ----------
cm = confusion_matrix(L, P, labels=list(range(NC)))
plt.figure(figsize=(10, 8)); sns.heatmap(np.log1p(cm), cmap="Blues", xticklabels=False, yticklabels=False, cbar_kws={"label": "log(1+count)"})
plt.title("Test confusion matrix, model C (102 classes)"); plt.xlabel("Predicted"); plt.ylabel("True")
plt.tight_layout(); plt.savefig(f"{OUT}/fig4_confusion_matrix.png", dpi=200); plt.show()
off = cm.copy(); np.fill_diagonal(off, 0); idx = np.dstack(np.unravel_index(np.argsort(-off.ravel())[:10], off.shape))[0]
pairs = pd.DataFrame([(int(i), int(j), int(off[i, j]), int(cm[i].sum())) for i, j in idx], columns=["true_class", "predicted_as", "count", "true_total"])
print("\nTop confused pairs (C):\n", pairs); pairs.to_csv(f"{OUT}/top_confused_pairs_C.csv", index=False)
rec_all = pd.DataFrame({k: [classification_report(out[k][1], out[k][0], labels=list(range(NC)), output_dict=True, zero_division=0)[str(c)]["recall"] for c in range(NC)] for k in "BC"}).round(4)
rec_all.to_csv(f"{OUT}/per_class_recall_B_C.csv"); print("\nWorst 10 classes (C recall):\n", rec_all.sort_values("C").head(10))

# ---------- Grad-CAM on the 4 most confident errors of C ----------
def gradcam(m, x, t):
    f = {}; m.eval()
    def fh(mod, i, o): f["a"] = o; o.register_hook(lambda gr: f.__setitem__("g", gr))
    hd = m.features[-1].register_forward_hook(fh)
    with torch.enable_grad(): o = m(x); m.zero_grad(); o[0, t].backward()
    hd.remove(); w = f["g"].mean((2, 3), keepdim=True); cam = torch.relu((w * f["a"]).sum(1, keepdim=True))
    cam = torch.nn.functional.interpolate(cam, size=(IMG, IMG), mode="bilinear")[0, 0]
    return (cam / (cam.max() + 1e-8)).detach().cpu().numpy()
conf = torch.softmax(Z, 1).max(1).values.numpy(); wrong = np.where(P != L)[0]; show = wrong[np.argsort(-conf[wrong])][:4]
fig, axs = plt.subplots(2, 4, figsize=(16, 8)); mis = []
for c, i in enumerate(show):
    axs[0, c].imshow(Xte[i]); axs[0, c].axis("off"); axs[0, c].set_title(f"True: class {L[i]} | Pred: class {P[i]} ({conf[i]:.2f})", fontsize=9)
    axs[1, c].imshow(Xte[i]); axs[1, c].imshow(gradcam(mods["C"], tf_plain(Xte[i]).unsqueeze(0).to(dev), int(P[i])), cmap="jet", alpha=.5); axs[1, c].axis("off")
    mis.append(dict(test_index=int(i), true=int(L[i]), pred=int(P[i]), confidence=float(conf[i])))
plt.tight_layout(); plt.savefig(f"{OUT}/fig5_misclassified_gradcam.png", dpi=200); plt.show()

# ---------- save, zip, download ----------
torch.save(mods["C"].state_dict(), f"{OUT}/efficientnet_b0_flowers102.pth")
info = dict(problem="PS-38 Flower Species Recognition", split=dict(train=len(Xtr), val=len(Xva), test=len(Xte)),
    test_per_class_min_max=[int(cte.min()), int(cte.max())], seed=SEED, batch_size=BS, device=str(dev),
    gpu=torch.cuda.get_device_name(0) if dev.type == "cuda" else "CPU", torch=torch.__version__,
    torchvision=torchvision.__version__, results=res, misclassified=mis)
json.dump(info, open(f"{OUT}/results_summary.json", "w"), indent=2)
print("\n===== SUMMARY =====\n", json.dumps(info, indent=2))
shutil.make_archive("/content/PS38_outputs", "zip", OUT)
from google.colab import files; files.download("/content/PS38_outputs.zip")
```
